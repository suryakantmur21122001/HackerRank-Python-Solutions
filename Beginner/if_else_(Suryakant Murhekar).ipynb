# **Problem: Python If-Else**

## Task Summary:
- If n is odd → print "Weird"
- If n is even and 2–5 → print "Not Weird"
- If n is even and 6–20 → print "Weird"
- If n is even and > 20 → print "Not Weird"
  
## Platform: HackerRank

```python
n = int(input().strip())
if n%2 !=0:
    print("Weird")
if n%2 ==0 and (2<=n<=5):
    print("Not Weird")
if n%2==0 and (6<=n<=20):
    print("Weird")
if n%2==0 and n>20:
    print("Not Weird")
```
